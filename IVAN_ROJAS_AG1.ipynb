##Devolución de cambio por técnica voraz

```
def cambio_monedas(N, SM):
  SOLUCION = [0]*len(SM)   #SOLUCION = [0,0,0,0,..]
  ValorAcumulado = 0

  for i,valor in enumerate(SM):
    monedas =  (N-ValorAcumulado)//valor
    SOLUCION[i] = monedas
    ValorAcumulado = ValorAcumulado + monedas*valor

    if ValorAcumulado == N:
      return SOLUCION


cambio_monedas(15,[25,10,5,1])
```

## N-Reinas por técnica de vueta atrás


```
def escribe(S):
  n = len(S)
  for x in range(n):
    print("")
    for i in range(n):
      if S[i] == x+1:
        print(" X " , end="")
      else:
        print(" - ", end="")


def es_prometedora(SOLUCION,etapa):
  #print(SOLUCION)
  #Si la solución tiene dos valores iguales no es valida => Dos reinas en la misma fila
  for i in range(etapa+1):
    #print("El valor " + str(SOLUCION[i]) + " está " +  str(SOLUCION.count(SOLUCION[i])) + " veces")
    if SOLUCION.count(SOLUCION[i]) > 1:
      return False

    #Verifica las diagonales
    for j in range(i+1, etapa +1 ):
      #print("Comprobando diagonal de " + str(i) + " y " + str(j))
      if abs(i-j) == abs(SOLUCION[i]-SOLUCION[j]) : return False
  return True



def reinas(N, solucion=[], etapa=0):
  if len(solucion) == 0:
      solucion=[0 for i in range(N)]

  for i in range(1, N+1):
    solucion[etapa] = i

    if es_prometedora(solucion, etapa):
      if etapa == N-1:
        print(solucion)
        #escribe(solucion)
        print()
      else:
        reinas(N, solucion, etapa+1)
    else:
      None

    solucion[etapa] = 0

reinas(8)
```

## Viaje por el rio. Programación dinámica

```
TARIFAS = [
[0,5,4,3,999,999,999],
[999,0,999,2,3,999,11],
[999,999, 0,1,999,4,10],
[999,999,999, 0,5,6,9],
[999,999, 999,999,0,999,4],
[999,999, 999,999,999,0,3],
[999,999,999,999,999,999,0]
]



################################################################
def Precios(TARIFAS):
################################################################
  #Total de Nodos
  N = len(TARIFAS[0])

  #Inicialización de la tabla de precios
  PRECIOS = [ [9999]*N for i in [9999]*N]
  RUTA = [ [""]*N for i in [""]*N]

  for i in range(0,N-1):
    RUTA[i][i] = i             #Para ir de i a i se "pasa por i"
    PRECIOS[i][i] = 0          #Para ir de i a i se se paga 0
    for j in range(i+1, N):
      MIN = TARIFAS[i][j]
      RUTA[i][j] = i

      for k in range(i, j):
        if PRECIOS[i][k] + TARIFAS[k][j] < MIN:
            MIN = min(MIN, PRECIOS[i][k] + TARIFAS[k][j] )
            RUTA[i][j] = k          #Anota que para ir de i a j hay que pasar por k
        PRECIOS[i][j] = MIN

  return PRECIOS,RUTA
################################################################


PRECIOS,RUTA = Precios(TARIFAS)
#print(PRECIOS[0][6])

print("PRECIOS")
for i in range(len(TARIFAS)):
  print(PRECIOS[i])

print("\nRUTA")
for i in range(len(TARIFAS)):
  print(RUTA[i])

#Determinar la ruta con Recursividad
def calcular_ruta(RUTA, desde, hasta):
  if desde == hasta:
    #print("Ir a :" + str(desde))
    return ""
  else:
    return str(calcular_ruta( RUTA, desde, RUTA[desde][hasta])) +  \
                ',' + \
                str(RUTA[desde][hasta] \
              )

print("\nLa ruta es:")
calcular_ruta(RUTA, 0,6)
```

```
import random

LISTA_1D = [random.randrange(1,10000) for x in range(1000)]

LISTA_2D = [(random.randrange(1,10000),random.randrange(1,10000)) for x in range(1000)]
```


#Primer intento: Fuerza bruta<br>


```
import math

def fuerza_bruta(lista):
    n = len(lista)
    if n < 2:
        return None, None, float('inf')

    p1 = lista[0]
    p2 = lista[1]
    dmin= abs(lista[0] - lista[1])

    for i in range(n):
        for j in range(i+1, n):
            d = abs(lista[i] - lista[j])
            if d < dmin:
                dmin = d
                p1 = lista[i]
                p2 = lista[j]

    return p1, p2, dmin

p1, p2, d = fuerza_bruta(LISTA_1D)

print(f"Puntos: {p1} y {p2}")
print(f"Distancia mínima: {d}")
```

#Calcular la complejidad. ¿Se puede mejorar?<br>

El bucle externo se ejecuta n veces.

El bucle interno, para cada i, se ejecuta desde i+1 hasta n-1, es decir, aproximadamente n/2 veces en promedio.

Complejidad: $n(n-1)/2 = O(n^2)$

Se puede mejorar con Divide y Vencerás, ya que al incrementar los datos, el algoritmo se vuelve lento.


#Segundo intento. Aplicar Divide y Vencerás<br>


```
def divide_y_venceras(lista):
    n = len(lista)
    if n <= 3:
        return fuerza_bruta(lista)
    m = n // 2
    izq= lista[:m]
    der = lista[m:]

    p1_izq, p2_izq, d_izq = divide_y_venceras(izq)
    p1_der, p2_der, d_der = divide_y_venceras(der)

    # Tomamos la mejor distancia de las mitades
    if d_izq < d_der:
       mejord, mejorp1, mejorp2 = d_izq, p1_izq, p2_izq
    else:
        mejord, mejorp1, mejorp2 = d_der, p1_der, p2_der

    # Verificar puntos cerca de la frontera
    medio = (izq[-1] + der[0]) / 2
    franja = []
    for x in lista:
      if abs(x - medio) < mejord:
        franja.append(x)

    # En la franja comparamos cada punto
    for i in range(len(franja)):
        for j in range(i+1, min(i+8, len(franja))):
            Distancia = abs(franja[i] - franja[j])
            if Distancia < mejord:
                mejord = Distancia
                mejorp1 = franja[i]
                mejorp2 = franja[j]

    return mejorp1, mejorp2, mejord

# Ordenar lista
LISTA_1D_O = sorted(LISTA_1D)

p1_dv, p2_dv, d_dv = divide_y_venceras(LISTA_1D_O)

print(f"Puntos: {p1_dv} y {p2_dv}")
print(f"Distancia mínima: {d_dv}")
```

#Calcular la complejidad. ¿Se puede mejorar?<br>

Cada vez que dividimos la lista a la mitad, el tamaño pasa de n a n/2, luego a n/4, luego a n/8, etc.

El proceso termina cuando llegamos a listas de tamaño 1 o 2.

Eso ocurre en log₂(n) pasos. Por ejemplo:

Con n = 8: 8 → 4 → 2 → 1, son 3 divisiones, log₂(8) = 3.

Con n = 16: 16 → 8 → 4 → 2 → 1, son 4 divisiones, log₂(16) = 4.

Número de niveles = log₂(n)

En cada nivel de la recursión, todos los subproblemas juntos manejan la misma cantidad total de datos: n.

Si en cada nivel se hace un trabajo de orden n, y hay log₂(n) niveles, entonces la complejidad es el producto.

Complejidad: $O(n×logn)$

La complejidad O(n log n) es la mejor posible para este problema, no se plantean mayores mejoras.


#Extender el algoritmo a 2D: [(1122, 6175), (135, 4076), (7296, 2741)…<br>


```
import math

def distancia(p1, p2):
    return math.sqrt((p1[0] - p2[0])**2 + (p1[1] - p2[1])**2)

def fuerza_bruta_2d(lista):
    n = len(lista)
    if n < 2:
        return None, None, float('inf')

    p1 = lista[0]
    p2 = lista[1]
    dmin = distancia(p1, p2)

    for i in range(n):
        for j in range(i+1, n):
            d = distancia(lista[i], lista[j])
            if d < dmin:
                dmin = d
                p1 = lista[i]
                p2 = lista[j]

    return p1, p2, dmin


def divide_y_venceras_2d(lista):
    n = len(lista)
    if n <= 3:
        return fuerza_bruta_2d(lista)

    m = n // 2
    izq = lista[:m]
    der = lista[m:]

    p1_izq, p2_izq, d_izq = divide_y_venceras_2d(izq)
    p1_der, p2_der, d_der = divide_y_venceras_2d(der)

    if d_izq < d_der:
        mejord, mejorp1, mejorp2 = d_izq, p1_izq, p2_izq
    else:
        mejord, mejorp1, mejorp2 = d_der, p1_der, p2_der

    # Franja con puntos cuya coordenada X está cerca del punto medio
    medio_x = (izq[-1][0] + der[0][0]) / 2
    franja = []
    for p in lista:
        if abs(p[0] - medio_x) < mejord:
            franja.append(p)

    # En la franja ordenamos y comparamos cada punto
    franja.sort(key=lambda p: p[1])
    for i in range(len(franja)):
        for j in range(i+1, len(franja)):
            if (franja[j][1] - franja[i][1]) >= mejord:
                break
            d = distancia(franja[i], franja[j])
            if d < mejord:
                mejord = d
                mejorp1 = franja[i]
                mejorp2 = franja[j]

    return mejorp1, mejorp2, mejord


LISTA_2D_O = sorted(LISTA_2D, key=lambda p: p[0])

p1, p2, d = fuerza_bruta_2d(LISTA_2D)
print(f"Fuerza bruta -> {p1}, {p2}, d = {d:.4f}")

p1, p2, d = divide_y_venceras_2d(LISTA_2D_O)
print(f"Divide y Vencerás -> {p1}, {p2}, d = {d:.4f}")
```

#Extender el algoritmo a 3D.<br>

```
def distancia_3d(p1, p2):
    return math.sqrt(
        (p1[0] - p2[0])**2 +
        (p1[1] - p2[1])**2 +
        (p1[2] - p2[2])**2
    )

def fuerza_bruta_3d(lista):
    n = len(lista)
    if n < 2:
        return None, None, float('inf')

    p1 = lista[0]
    p2 = lista[1]
    dmin = distancia_3d(p1, p2)

    for i in range(n):
        for j in range(i+1, n):
            d = distancia_3d(lista[i], lista[j])
            if d < dmin:
                dmin = d
                p1 = lista[i]
                p2 = lista[j]

    return p1, p2, dmin

def divide_y_venceras_3d(lista):
    n = len(lista)
    if n <= 3:
        return fuerza_bruta_3d(lista)

    m = n // 2
    izq = lista[:m]
    der = lista[m:]

    p1_izq, p2_izq, d_izq = divide_y_venceras_3d(izq)
    p1_der, p2_der, d_der = divide_y_venceras_3d(der)

    if d_izq < d_der:
        mejord, mejorp1, mejorp2 = d_izq, p1_izq, p2_izq
    else:
        mejord, mejorp1, mejorp2 = d_der, p1_der, p2_der

    # Franja: puntos cuya X está cerca del punto medio
    medio_x = (izq[-1][0] + der[0][0]) / 2
    franja = []
    for p in lista:
        if abs(p[0] - medio_x) < mejord:
            franja.append(p)

    # Ordenamos por Y y comparamos
    franja.sort(key=lambda p: p[1])
    for i in range(len(franja)):
        for j in range(i+1, len(franja)):
            if (franja[j][1] - franja[i][1]) >= mejord:
                break
            d = distancia_3d(franja[i], franja[j])
            if d < mejord:
                mejord = d
                mejorp1 = franja[i]
                mejorp2 = franja[j]

    return mejorp1, mejorp2, mejord

#Generamos datos para el 3D
LISTA_3D = [(random.randrange(1, 10000),
             random.randrange(1, 10000),
             random.randrange(1, 10000)) for _ in range(1000)]

LISTA_3D_O = sorted(LISTA_3D, key=lambda p: p[0])

p1_fb, p2_fb, d_fb = fuerza_bruta_3d(LISTA_3D)
print(f"Fuerza bruta -> {p1_fb}, {p2_fb}, d = {d_fb:.4f}")

p1, p2, d = divide_y_venceras_3d(LISTA_3D_O)
print(f"Divide y Vencerás -> {p1}, {p2}, d = {d:.4f}")
```
